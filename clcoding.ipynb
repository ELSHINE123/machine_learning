```python
#Find the minimum and maximum in the range of values
numbers = [40, 10, 35, 5, 90, 20, 30, 56, 70, 12, 80]
```

```python
numbers
```

```python
largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print(largest)
```

```python
smallest = numbers[0]

for number in numbers:
    if number < smallest:
        smallest = number

print(smallest)
```

```python
fruits = ("Apple Cherry Guava Apple Guava Apple").split()
```

```python
fruits
```

```python
frequency = {}

for fruit in fruits:
    if fruit in frequency:
        frequency[fruit] += 1
    else:
        frequency[fruit] = 1

print(frequency)
```

```python
#Transforming a List
numbers = [1,2,3,4,5]

squares = []

for num in numbers:
    squares.append(num ** 2)
    
print(squares)
```

```python
#List Comprehension. [Expression for item in Iterable]
squares1 = [number ** 2 for number in numbers]
print(squares1)
```

```python
fruits_lower = [fruit.lower() for fruit in fruits]
print(fruits_lower)
```

```python
# Select only even numbers. Filtering with list comprehension
numbs = [1,2,3,4,5,6,7,8,99,32,45,22,35,65,45,36,32,34,56]

even_nums = [num for num in numbs if num % 2 == 0]

print(even_nums)
```

```python
even_squares = [num ** 2 for num in numbs if num % 2 == 0]
print(even_squares)
```

```python
sqrs = {}

for num in numbs:
    sqrs[num] = num ** 2

print(sqrs)
```

```python
sqrs2 = {
    num: num ** 2 for num in numbs}
sqrs2
```
